```python
# np.meshgrid()的使用
import numpy as np
# 打印全部内容
np.set_printoptions(threshold=np.inf) # 设置打印选项，threshold=np.inf表示打印所有内容，不进行省略显示
np.random.seed(0) # 设置随机种子，确保每次运行结果相同
x = np.random.rand(10) * 10 # 生成10个0到10之间的随机数作为x坐标
y = np.random.rand(10) * 10 # 生成10个0到10之间的随机数作为y坐标
X, Y = np.meshgrid(x, y) # np.meshgrid()函数用于生成网格坐标矩阵，其原理是将输入的一维数组x和y分别复制成二维矩阵X和Y，其中X的每一行都是x的副本，Y的每一列都是y的副本，这样就形成了一个网格坐标矩阵
# 网格坐标矩阵能够方便地进行函数值的计算和可视化，例如在绘制等高线图或三维曲面图时，常常需要使用网格坐标矩阵来计算函数值并进行绘制。
print('X.shape:', X.shape) # 打印X的形状
print('Y.shape:', Y.shape) # 打印Y的形状
print('X:', X) # 打印X的值
print('Y:', Y) # 打印Y的值
```
