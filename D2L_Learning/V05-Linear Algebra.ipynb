# Linear Algebra

```python
import torch

# 向量的创建 
# 向量可以被视为标量值组成的列表。 这些标量值被称为向量的元素（element）或分量（component）
x = torch.arange(4)
print('x:', x)

print('x.shape:', x.shape) # 向量的形状
print('x.size:', x.size()) # 向量的大小
print('len(x):', len(x)) # 向量的长度
print('x.dtype:', x.dtype) # 向量的数据类型
print('x.numel:', x.numel()) # 向量的元素数量


# 矩阵的创建
# 矩阵是一个二维数组。 可以看做向量的推广
A = torch.arange(20).reshape(5, 4)
print('A:\n', A)
print('A.shape:', A.shape) # 矩阵的形状
print('A.size:', A.size()) # 矩阵的大小
print('len(A):', len(A)) # 矩阵的行数
print('A.dtype:', A.dtype) # 矩阵的数据类型
print('A.numel:', A.numel()) # 矩阵的元素数量
print('A^T:\n', A.T) # 矩阵的转置

# 张量的创建
# 张量是一个多维数组。 可以看做矩阵的推广
B = torch.arange(24, dtype=torch.float32).reshape(2, 3, 4)
print('tensor B:\n', B)
print('B.shape:', B.shape) # 张量的形状
print('B.size:', B.size()) # 张量的大小
print('len(B):', len(B)) # 张量的行数
```

## calculation

```python
# calculation
A1 = A.clone() # 克隆矩阵A，得到一个新的矩阵A1 
# 通过分配新内存，将A的一个副本分配给A1, 使得A1和A不共享内存，修改A1不会影响A

print('A * A1:\n', A * A1) # 矩阵的元素级乘法（哈达玛积），得到一个新的矩阵，每个元素是A和A1对应元素的乘积

# 通过指定axis参数来计算矩阵或张量的元素和
A = torch.arange(20).reshape(5, 4)
print('matrix A:\n', A)
B = torch.arange(24, dtype=torch.float32).reshape(2, 3, 4)
print('tensor B:\n', B)
print('A.sum():', A.sum()) # A的所有元素的和，得到一个标量值
print('A.sum(axis=0):', A.sum(axis=0)) # 沿着行的方向求和，得到一个新的矩阵，每个元素是A的每列元素的和
print('A.sum(axis=1):', A.sum(axis=1)) # 沿着列的方向求和，得到一个新的矩阵，每个元素是A的每行元素的和
print('B.sum():', B.sum()) # B的所有元素的和，得到一个标量值
print('B.sum(axis=0):\n', B.sum(axis=0)) # 沿着第0维求和，得到一个新的矩阵，每个元素是B的每个3x4矩阵同位置元素的和
print('B.sum(axis=1):\n', B.sum(axis=1)) # 沿着第1维求和，得到一个新的矩阵，每个元素是B的每个2x4矩阵同位置元素的和
print('B.sum(axis=2):\n', B.sum(axis=2)) # 沿着第2维求和，得到一个新的矩阵，每个元素是B的每个2x3矩阵同位置元素的和
print('B.sum(axis=-1):\n', B.sum(axis=-1)) # 沿着最后一个维度求和，得到一个新的矩阵，每个元素是B的每个2x3矩阵同位置元素的和
print('B.sum(axis=[0,1]):\n', B.sum(axis=[0,1] )) # 沿着第0和第1维求和，得到一个新的矩阵，每个元素是B的每个4矩阵同位置元素的和


# 平均值相关
print('A.mean():', A.float().mean()) # A的所有元素的平均值，得到一个标量值
print('A.mean(axis=0):', A.float().mean(axis=0)) # 沿着行的方向求平均值，得到一个新的矩阵，每个元素是A的每列元素的平均值
print('B.mean(axis=2):', B.float().mean(axis=2)) # 沿着第2维求平均值，得到一个新的矩阵，每个元素是B的每个2x4矩阵元素的平均值
print('B.mean(axis=[0,1]):', B.float().mean(axis=[0,1])) # 沿着列的方向求平均值，得到一个新的矩阵，每个元素是A的每行元素的平均值

# 计算总和或均值时保持轴数不变 
print('A.sum(axis=1):', A.sum(axis=1)) 
print('A.sum(axis=1, keepdims=True):', A.sum(axis=1, keepdims=True)) # 沿着行的方向求和，得到一个新的矩阵，每个元素是A的每列元素的和，保持轴数不变
print('B.mean(axis=2):', B.mean(axis=2))
print('B.mean(axis=2, keepdims=True):', B.mean(axis=2, keepdims=True)) # 沿着第2维求平均值，得到一个新的矩阵，每个元素是B的每个24矩阵元素的平均值，保持轴数不变
# 我们可以通过广播将A除以sum_A。 由于sum_A的轴数与A不同，我们需要保持sum_A的轴数不变，以便进行广播运算。
sum_A = A.sum(axis=1, keepdims=True)
print('A:\n', A)
print('A / sum_A:\n', A / sum_A) # A的每个元素除以sum_A的对应元素，得到一个新的矩阵，每个元素是A的每列元素除以A的每行元素的和
print('B:\n', B)
print('B / B.sum(axis=2, keepdims=True):\n', B / B.sum(axis=2, keepdims=True)) # 得到一个新的矩阵，每个元素是B的每个2x4矩阵元素除以B的每个2x4矩阵元素的和

# 某个轴上的累积和
print('A.cumsum(axis=0):\n', A.cumsum(axis=0)) # 得到一个新的矩阵， 每个元素是A的每列元素上自上而下到目前位置的累加和
print('A.cumsum(axis=1):\n', A.cumsum(axis=1)) # 得到一个新的矩阵， 每个元素是A的每行元素上自左而右到目前位置的累加和
print('B.cumsum(axis=2):\n', B.cumsum(axis=2)) # 沿着第2维计算累积和，得到一个新的矩阵，每个元素是B的每个2x4矩阵元素的累积和
```

```python
import numpy as np

# 点积运算
# 向量之间
x = np.arange(4)
y = np.ones(4)
print(np.dot(x, y)) # 向量点积的代数运算，逐项相乘再求和，得到一个标量值
print(x @ y) # 等价于np.dot(x, y)

# 向量与矩阵之间
A = np.arange(20).reshape(5, 4)
print('A:\n', A)
print('x:\n', x)
print('A.dot(x):\n', A.dot(x)) # 矩阵与向量的点积，得到一个新的向量，每个元素是A的每行元素与x的点积
print('A @ x:\n', A @ x) # 矩阵与向量的点积，得到一个新的向量，每个元素是A的每行元素与x的点积

B = np.arange(24).reshape(4, 6)
print('A:\n', A)
print('B:\n', B)
print('np.dot(A, B):\n', np.dot(A, B)) # 矩阵乘法，得到一个新的矩阵，每个元素是A的每行元素与B的每列元素的点积
print('A @ B:\n', A @ B) # 矩阵乘法，得到一个新的矩阵，每个元素是A的每行元素与B的每列元素的点积


```

### (Norm Calculation) 范数计算

```python

import torch

# 向量范数
u1 = torch.arange(4, dtype=torch.float32)
print('L1 norm of u1 (manual):', torch.abs(u1).sum()) # 手动计算L1范数，得到一个标量值
print('L1 norm of u1 (torch):', torch.norm(u1, p=1)) # L1范数，得到一个标量值，表示u1的绝对值之和

u2 = torch.tensor([3.0, -4.0])
print('L2 norm of u2:', torch.norm(u2)) # L2范数，得到一个标量值，表示u2的欧几里得长度

u3 =torch.randn_like((u1))
print('u3:', u3)
print('LInf norm of u3:', torch.norm(u3, p=float('inf'))) # L∞范数，得到一个标量值，表示u3的最大绝对值元素


# 矩阵范数
U1 = torch.arange(20, dtype=torch.float32).reshape(5, 4)
print('matrix U1:\n', U1)
print('Frobenius norm of U1:', torch.norm(U1)) # Frobenius范数，得到一个标量值，表示U1的元素平方和的平方根

# 矩阵的L1诱导范数
print('L1 induced norm of U1:', torch.linalg.norm(U1, ord=1)) # L1诱导范数，得到一个标量值，表示U1的列绝对值和的最大值

# 矩阵的L∞诱导范数
print('LInf induced norm of U1:', torch.linalg.norm(U1, ord=float('inf'))) # L∞诱导范数，得到一个标量值，表示U1的行绝对值和的最大值

# 矩阵的谱范数（L2诱导范数）
print('Spectral norm of U1:', torch.linalg.norm(U1, ord=2)) # 谱范数，得到一个标量值，表示U1的最大奇异值
# 求矩阵的奇异值
print('Singular values of U1:', torch.linalg.svdvals(U1)) # 矩阵的奇异值，得到一个向量，每个元素是U1的一个奇异值
# 矩阵的核范数（L2诱导范数的对偶范数）
print('Nuclear norm of U1:', torch.linalg.norm(U1, ord='nuc')) # 核范数，得到一个标量值，表示U1的奇异值之和
```
