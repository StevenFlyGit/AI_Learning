# Data Operation

## Write information into csv Document

```python
import os

os.makedirs(os.path.join('..', 'data'), exist_ok=True)
data_file = os.path.join('..', 'data', 'house_tiny.csv')
with open(data_file, 'w') as f:
    f.write('NumRooms,Alley,Price\n')  # 列名
    f.write('NA,Pave,127500\n')  # 每行表示一个数据样本
    f.write('2,NA,106000\n')
    f.write('4,Start,178100\n')
    f.write('NA,NA,140000\n')
```

```python
# 如果没有安装pandas，只需取消对以下行的注释来安装pandas
# !pip install pandas
import pandas as pd

data = pd.read_csv(data_file)
print(data)
```

## 处理缺失数据
“NaN”项代表缺失值

```python
inputs, outputs = data.iloc[:, 0:2], data.iloc[:, 2]
print('inputs:\n', inputs)
print('outputs:\n', outputs)

try:
    inputs = inputs.fillna(inputs.mean()) # 尝试对所有列按均值填充缺失值，但对于字符串列会报错
    print(inputs)
except Exception as e:
    print("Error filling missing values:", e)

inputs = inputs.fillna(inputs.mean(numeric_only=True))   # 仅对数值列按均值填充
print(inputs)
inputs = pd.get_dummies(inputs, dummy_na=True)  # 将类别列和缺失值转换为指示列，一个值为一个类别，类似于sql中的group by操作
print(inputs)
```

```python
import torch

X = torch.tensor(inputs.to_numpy(dtype=float))
y = torch.tensor(outputs.to_numpy(dtype=float))
# 转换为张量格式，以便后续的机器学习模型可以使用，传统机器学习默认用float64，深度学习默认用float32
print(X)
print(y)
```
