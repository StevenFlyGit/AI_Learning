## Operation related to tensor - 有关张量操作

### Tensor.reshape() 方法 - 改变张量形状 - 注意-1的用法

````python
# Tensor.reshape() 是 PyTorch 中用于改变张量形状的核心方法，不改变数据内容，只改变视图。4

import torch

x = torch.arange(12)  # [0, 1, 2, ..., 11]，shape: (12,)
print(f"原始: {x}")

# reshape 为 3行4列
y = x.reshape(3, 4)
print(f"变形后: {y}")
print(y)
````

| 规则          | 说明                         |
| :---------- | :------------------------- |
| **元素总数不变**  | 新形状的元素个数必须等于原形状            |
| **-1 自动推断** | 用 `-1` 让 PyTorch 自动计算该维度大小 |
| **共享内存**    | reshape 后通常共享原数据内存（视图操作）   |




````python
# 1. 自动推断维度（-1）
x = torch.randn(12, 10)  # 120 个元素
print(f"原始1: {x}")

# 只知道要变成 3 行，列数自动算：120/3 = 40
y = x.reshape(3, -1)      # shape: (3, 40)
print(f"变形后1: {y}")

# 只知道要 5 列，行数自动算
z = x.reshape(-1, 5)      # shape: (24, 5)
print(f"变形后2: {z}")

# 变成 1D 向量
flat = x.reshape(-1)       # shape: (120,)
print(f"变形后3: {flat}")


# 2. 多维 reshape
x = torch.randn(2, 3, 4)   # shape: (2, 3, 4) = 24 个元素
print(f"原始2: {x}")

# 各种变形方式
y1 = x.reshape(2, 12)      # (2, 12)
print(f"变形后4: {y1}")
y2 = x.reshape(6, 4)       # (6, 4)
print(f"变形后5: {y2}")
y3 = x.reshape(2, 3, 2, 2) # (2, 3, 2, 2)
print(f"变形后6: {y3}")
y4 = x.reshape(1, 2, 3, 4) # 增加维度


x = torch.tensor(5.0)      # shape: ()
y = x.reshape(1)           # shape: (1,)，变成 1 元素向量


# 3. 0 维（标量）处理
x = torch.tensor(5.0)      # shape: ()
y = x.reshape(1)           # shape: (1,)，变成 1 元素向量
````

| 方法            | 作用   | 关键区别            |
| :------------ | :--- | :-------------- |
| `reshape()`   | 改变形状 | **最灵活**，必要时拷贝数据 |
| `view()`      | 改变形状 | 要求内存连续，否则报错     |
| `permute()`   | 交换维度 | 只换轴顺序，不改变形状逻辑   |
| `transpose()` | 转置   | 交换两个指定维度        |


````python
x = torch.randn(2, 3, 4)

# reshape: 可以任意改变形状
y = x.reshape(3, 2, 4)     # OK，重新排列

# view: 要求连续内存
y = x.view(3, 2, 4)        # 可能报错 if x 不连续

# permute: 只换轴顺序
y = x.permute(1, 0, 2)     # shape: (3, 2, 4)，轴 0 和 1 交换


# 增加/减少维度
x = torch.randn(3, 4)      # (3, 4)
print('x: \n', x)

# 增加 batch 维度
y = x.reshape(1, 3, 4)     # (1, 3, 4)，或 x.unsqueeze(0)
print(x.unsqueeze(2))
print('y: \n', y)
y2 = x.reshape(-1, 3)      # (4, 3)，自动推断行数
print('y2: \n', y2)

# 减少维度（单元素）
z = torch.tensor([[5.0]])  # (1, 1)
print('z: \n', z)
s = z.reshape(())          # ()，标量，或 z.item()
print('s: \n', s)
````

### unsqueeze() - 在指定位置插入一个大小为 1 的新维度

````python
import torch

x = torch.randn(3, 4)      # shape: (3, 4)

# 在第 0 维前插入新维度
y0 = x.unsqueeze(0)        # shape: (1, 3, 4)
print('y0: \n', y0)

# 在第 1 维前插入新维度  
y1 = x.unsqueeze(1)        # shape: (3, 1, 4)
print('y1: \n', y1)

# 在第 2 维前插入新维度（最后）
y2 = x.unsqueeze(2)        # shape: (3, 4, 1)
print('y2: \n', y2)

# 负数索引：-1 表示最后一维之后
y_neg = x.unsqueeze(-1)      # shape: (3, 4, 1)，等价于 unsqueeze(2)


# 场景1：单样本增加 batch 维度
image = torch.randn(3, 224, 224)      # (C, H, W)
batch = image.unsqueeze(0)            # (1, C, H, W)

# 场景2：为广播机制准备
a = torch.randn(3, 4)
print('a: \n', a)
b = torch.randn(4)
print('b: \n', b)
# a * b 报错，需要：
b = b.unsqueeze(0)                    # (1, 4)，现在可以广播
print('b unsqueeze: \n', b)
print ('a * b: \n', a * b)

# 场景3：链式调用增加多维度
x = torch.randn(3)
y = x.unsqueeze(0).unsqueeze(2)         # (1, 3, 1)
````

| 操作             | 写法                   | 结果          |
| :------------- | :------------------- | :---------- |
| `unsqueeze(0)` | `x.unsqueeze(0)`     | `(1, 3, 4)` |
| 等价 reshape     | `x.reshape(1, 3, 4)` | `(1, 3, 4)` |
| `unsqueeze(1)` | `x.unsqueeze(1)`     | `(3, 1, 4)` |
| 等价 reshape     | `x.reshape(3, 1, 4)` | `(3, 1, 4)` |


### squeeze() - 指定位置移除大小为1的维度

````python
x = torch.randint(0, 100, (1,2)) 
print('x: \n', x)

x = torch.randn(1, 3, 1, 4, 1)
print('x: \n', x)

# 去掉所有大小为1的维度
y = x.squeeze()           # shape: (3, 4)
print('y: \n', y)

# 去掉指定维度（该维必须为1）
y0 = x.squeeze(0)         # shape: (3, 1, 4, 1)
print(y0.shape)
print('y0: \n', y0)
y2 = x.squeeze(2)         # shape: (1, 3, 4, 1)
print('y2: \n', y2)
y3 = x.squeeze(4)         # shape: (1, 3, 1, 4)
print('y3: \n', y3)
````
